```python
# jajan dikantin
#definisikan variabel

harga_gorengan = 2500
jumlah_gorengan = 3
harga_esteh = 4000
jumlah_esteh = 2
uang = 20000

#total belanja
total = harga_gorengan * jumlah_gorengan + harga_esteh * jumlah_esteh

#uang kembalian
kembalian = uang - total

#cukup atau tidak
cukup = uang >= total

print("total:",total)
print("kembalian:", kembalian)
print("Uang cukup?:", cukup)
```

```python
#konversi ke fungsi input () untuk meminta masukan dari user
harga_gorengan = int(input("masukan harga gorengan:"))
jumlah_gorengan = int(input("masukan jumlah gorengan:"))
harga_esteh = int(input("masukan harga esteh:"))
jumlah_esteh = int(input("masukan jumlah esteh:"))
uang = int(input("masukan uang:"))

#total belanja
total = harga_gorengan * jumlah_gorengan + harga_esteh * jumlah_esteh

#uang kembalian
kembalian = uang - total

#cukup atau tidak
cukup = uang >= total

print("total:",total)
print("kembalian:", kembalian)
print("Uang cukup?:", cukup)
```

```python
# durasi dalam menit
total_menit = 275
# jam penuh
jam_penuh = total_menit // 60
# sisa menit
menit_sisa = total_menit % 60
# jam desimal
jam_desimal = total_menit / 60
# cek apakah durasi lebih dari 3 jam kurang dari 5 jam
dalam_rentang = total_menit >= 180 and total_menit <= 300

print("Durasi :", jam_penuh, "jam_penuh", menit_sisa, "total_menit")
print("Jam desimal ;", round(jam_desimal, 2))
print("3 s.d. 5 jsm?:", dalam_rentang)
```
