```python
# 1. Operasi +, -, *, /, //, %

bilangan1 = int(input("Masukkan bilangan pertama: "))
bilangan2 = int(input("Masukkan bilangan kedua: "))

print(f"Hasil penjumlahan: {bilangan1 + bilangan2}")
print(f"Hasil pengurangan: {bilangan1 - bilangan2}")
print(f"Hasil perkalian: {bilangan1 * bilangan2}")
print(f"Hasil pembagian: {bilangan1 / bilangan2}")
print(f"Hasil pembagian bulat: {bilangan1 // bilangan2}")
print(f"Hasil modulus: {bilangan1 % bilangan2}")
```

```python
# 2. Operator Relasional

nilai = int(input("Masukkan nilai ujian (0-100): "))

lulus = nilai >= 60

print(f"Apakah nilai ini termasuk kategori lulus (>= 60)? {lulus}")
```

```python
# 3. umur dan SIM

umur = int(input("Masukkan umur: "))
punya_sim = input("Apakah memiliki SIM? (True/False): ").strip().lower() == "true"

boleh_mengemudi = (umur >= 17) and punya_sim

print(f"Apakah boleh mengemudi sendiri? {boleh_mengemudi}")
```

```python
# 4. efisiensi termal

energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi termal pembangkit adalah {efisiensi:.2f}%")
```

```python
# 5. kondisi maintenance pembangkit

suhu_boiler = float(input("Masukkan suhu boiler (°C): "))
beban_operasi = float(input("Masukkan beban operasi (%): "))

perlu_maintenance = (suhu_boiler > 550) or (beban_operasi > 95)

print(f"unit pembangkit dengan suhu boiler {suhu_boiler}°C dan beban operasi {beban_operasi}%, "
f"status perlu maintenance: {perlu_maintenance}")
```
