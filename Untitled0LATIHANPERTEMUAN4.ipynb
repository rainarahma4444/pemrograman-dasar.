```python
# Latihan pertemuan 3 dan 4
# RAINA RAHMA
# 4232501013

# soal nomor 1 (studi kasus)
# Sebuah tahun merupakan tahun kabisat jika memnuhi satu dari dua kondisi berikut
# a. tahun tersebut habis dibagi 4 tetapi tidak habis dibagi 100
# b. tahun tersebut habis dibagi 400
# buatlah sebuah ekspresi logika kompleks untuk menentukan apakah suatu tahun merupakan tahun kabisat

tahun = int(input("Masukkan tahun: "))
tahun_kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)

print("Apakah tahun", tahun, "merupakan tahun kabisat?", tahun_kabisat)
```

```python
# soal nomor 2
# buatlah progam sederhana versi kalian masing masing dengan lingkup, fungsi input, fungsi output dan operasi aritmatika

nama = input("Masukkan nama: ")
daya = float(input("Masukkan daya alat listrik (Watt): "))
lama_pemakaian = float(input("Masukkan lama pemakaian (jam): "))

energi = (daya * lama_pemakaian) / 1000

print("\n--- HASIL PERHITUNGAN ---")
print("Nama:", nama)
print("Daya alat:", daya, "Watt")
print("Lama pemakaian:", lama_pemakaian, "jam")
print("Energi yang digunakan:", energi, "kWh")
```
