```python
# soal 1 Daya Generator dan Efisiensi Turbin
# a. hitung daya listrik
tegangan = 220  # Volt
arus = 15  # Ampere
daya_listrik = tegangan * arus

print(f"Daya listrik = {daya_listrik} W")

# b. Menyimpan energi input dan output serta menghitung efisiensi turbin
energi_input = 500
energi_output = 400
efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi = {efisiensi:.2f}%")

# PENJELASAN  :
# untuk mendapatkan nilai daya listrik menggunakan operator perkalian dengan mengalikan nilai dari variabel tegangan dan arus
# untuk mendapatkan nilai efisiensi yaitu dengan membagi antara nilai output dan nilai input
```

```python
# soal 3 data suhu boiler
# Data awal suhu boiler
suhu = [92.5, 88.7, 95.1, 102.3, 97.8]

# a. suhu pada titik ke-3 (indeks 2)
print(f"Suhu pada titik ke-3 = {suhu[2]} °C")

# b. tiga pembacaan pertama menggunakan slicing (indeks 0 hingga 2)
print(f"Tiga pembacaan pertama = {suhu[:3]}")

# c. Menambahkan pembacaan baru 99.4 ke dalam list dan menampilkan jumlah data
suhu.append(99.4)
print(f"Jumlah data setelah penambahan = {len(suhu)}")

# PENJELASAN :
# suhu titik ke 3 indeks list dengan ururtan 0 1 2 3 4 sehingga suhu ke 3 dengan urutan ke 2 adalah 95.1  °C
# Append & Len menambahkan item baru sehingga totalnya ada 6
```

```python
# soal 5 katup pengaman
# Membaca tekanan_uap dari input (bertipe float)
tekanan_uap = float(input("Masukkan tekanan uap (psi): "))

# Status manual override (diberikan langsung dalam kode, silakan ubah untuk pengujian)
status_manual_override = False

# Menentukan status katup menggunakan if-elif-else
if status_manual_override == True:
  print("Status: Katup Mode Manual")
elif tekanan_uap > 150 and status_manual_override == False:
  print("Status: Katup Terbuka")
else:
  print("Status: Katup Tertutup")

# PENJELASAN :
# input tekanan uap yang dimasukan adalah 140 merupakan nilai yang kurang dari 150 maka letak nya pada percabangan else dimana status katup tertutup
# tetapi jika memasukkan angka input diatas 150 maka akan masuk ke percabangan elif yaitu status katup terbuka
```

```python
# soal 7 pemanas mencapai target
suhu_saat_ini = 60
target_suhu = 85

while suhu_saat_ini < target_suhu:
  suhu_saat_ini += 5
  print(f"Suhu saat ini: {suhu_saat_ini} C")

  if suhu_saat_ini == 80:
    print("WARNING: Mendekati target!")

print(f"Target {target_suhu} C tercapai.")

# PENJELASAN :
# perulangan while akan terus berjalan sampai batas tertentu atau di bawah angka target yaitu 85
# terdapat if untuk melakukan pengecekan jika suhu nya menyentuh 80 progam akan mencetakn WARNING
```

```python
# soal 9 pembacaan vibrasi dengan sensor rusak
data = ["1.2", "0.9", "ERR", "3.1", "1.8"]
jumlah_tidak_valid = 0

for i, val in enumerate(data, start=1):
  try:
    # a. Ubah data menjadi float
    nilai = float(val)
  except ValueError:
    # Jika gagal (ValueError), tangani dan lanjutkan ke unit berikutnya
    print(f"Unit {i}: data tidak valid")
    jumlah_tidak_valid += 1
    continue

  # b. Jika nilai melebihi 2.0
  if nilai > 2.0:
    print(f"Warning: Unit {i} vibrasi tinggi!")

  # c. Jika nilai melebihi 3.0, hentikan perulangan
  if nilai > 3.0:
    print("KRITIS: pemeriksaan dihentikan")
    break

# d. Setelah perulangan selesai, tampilkan jumlah data tidak valid
print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")

# PENJELASAN :
# terdapat try-except supay kalau datanya gagal diubah jadi angka, program menangkap error yang muncul dan melamjutkannya ke pengecekan dengan continue
```
