```python
# SOAL OPERATOR PADA PYTHON MINGGU KE 5
# Soal 1 : Belanja Buku
# Budi membeli 7 buku dengan harga Rp 18.500 per buku. Toko memberi diskon 10% dari total belanja. Budi membayar dengan uang Rp 150.000.
# Tentukan
Harga_buku = 18500
Jumlah_buku = 7
Uang = 150000

#total belanja
total = Harga_buku * Jumlah_buku
#diskon 10%
diskon = total * 10 // 100
#jumlah yang harus dibayar
bayar = total - diskon
#jumlah kembalian
kembalian = Uang - bayar
#iuran 4 orang
per_orang = bayar // 4
sisa = bayar % 4
#uang cukup atau tidak
cukup = Uang >= bayar


print("Total      :", total)
print("Diskon     :", diskon)
print("Bayar      :", bayar)
print("Kembalian  :", kembalian)
print("Per orang  :", per_orang)
print("Sisa       :", sisa)
print("Uang cukup?:", cukup)
```

```python
# SOAL 2 : PLTS Atap rumah
# Sebuah rumah memasang 12 panel surya berdaya 450 Wp per panel. Rata-rata penyinaran efektif 5 jam/hari dengan efisiensi sistem 80% (0,8).
# Kebutuhan listrik rumah 25 kWh/hari dan tarif listrik Rp 1.444,70/kWh.
# Tentukan
Jumlah_panel = 12
Daya_panel = 450 / 1000 # ubah dari wp ke KW
Penyinaran_efektif = 5
Efisiensi = 0.8
Kebutuhan = 25
Tarif_listrik = 1444.70

# energi yang dihasilkan per hari
Energi_harian = round(Jumlah_panel * Daya_panel * Penyinaran_efektif * Efisiensi, 1)
# energi per bulan
Energi_bulanan = round(Energi_harian * 30, 1)
# nilai listrik yang dihasilkan
Nilai_listrik = round(Energi_bulanan * Tarif_listrik, 1)
# kekurangan energi per hari
Kekurangan = round(Kebutuhan - Energi_harian, 1)
# apakah mencukupi kebutuhan?
Mencukupi = Energi_harian >= Kebutuhan

print("Energi harian (kWh) :", Energi_harian)
print("Energi bulanan (kWh):", Energi_bulanan)
print("Nilai listrik (Rp)  :", Nilai_listrik)
print("Kekurangan (kWh)    :", Kekurangan)
print("Mencukupi?          :", Mencukupi)
```

```python
# SOAL 3 : Generator Diesel dan Rugi Kabel
# Sebuah generator diesel 200 kW beroperasi 8 jam/hari. Konsumsi bahan bakar 0,25 liter per kWh, harga solar Rp 13.500/liter, dan kapasitas tangki 1.000 liter.
# Kabel penghantar dialiri arus 20 A dengan hambatan 0,5 Ω.
# Tentukan
Daya = 200
Jam = 8
Konsumsi = 0.25
Harga_solar = 13500
Kapasitas_tangki = 1000
Arus = 20
Hambatan = 0.5

# Energi yang dibangkitkan per hari
Energi = Daya * Jam
# Kebutuhan solar dan biaya
Solar = Energi * Konsumsi
Biaya = Solar * Harga_solar
# Kebutuhan solar dan biaya
Tangki_cukup = Kapasitas_tangki // Solar
Sisa_solar = Kapasitas_tangki % Solar
# Rugi daya pada kabel
Rugi_kabel = Arus ** 2 * Hambatan
# Apakah rugi kabel tinggi?
Rugi_tinggi = Rugi_kabel > 150 and Arus >= 20

print("Energi (kWh)      :", Energi)
print("Solar (liter)     :", Solar)
print("Biaya (Rp)        :", Biaya)
print("Tangki cukup      :", Tangki_cukup, "hari")
print("Sisa solar (L)    :", Sisa_solar)
print("Rugi kabel (W)    :", Rugi_kabel)
print("Rugi tinggi?      :", Rugi_tinggi)
```
