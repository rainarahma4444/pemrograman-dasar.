```python
# Pola akumulator: total, rata-rata, hitungan ( AKTIVITAS 6D)

skor = [80, 65, 90, 75, 70]

total = 0                    # akumulator: mulai dari nol
jumlah_lulus = 0             # penghitung = akumulator yang menambah 1
for nilai in skor:
    total = total + nilai    # tambahkan setiap elemen
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total sementara:", total)

rata_rata = total / len(skor)
print("Total     :", total)
print("Rata-rata :", rata_rata)
print("Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

```python
# Nilai tertinggi dan terendah

skor = [80, 65, 90, 75, 70]

tertinggi = skor[0]          # terbaik sejauh ini (anggap yang pertama)
terendah = skor[0]
for nilai in skor:
    if nilai > tertinggi:
        tertinggi = nilai
    if nilai < terendah:
        terendah = nilai

print("Tertinggi:", tertinggi)
print("Terendah :", terendah)
print("Cek fungsi bawaan:", max(skor), min(skor))
```

```python
# Bukan hanya nilainya, tetapi posisinya

suhu = [28, 31, 30, 33, 29]      # suhu (°C), hari ke-1 sampai ke-5

pos_maks = 0                      # posisi terbaik sejauh ini
for i in range(len(suhu)):
    if suhu[i] > suhu[pos_maks]:
        pos_maks = i              # simpan POSISI, bukan hanya nilai

print("Suhu tertinggi:", suhu[pos_maks], "°C")
print("Ada di indeks", pos_maks, "= hari ke-", pos_maks + 1)
```

```python
# Pencarian linear: berhenti di kecocokan pertama

suhu = [28, 31, 30, 33, 29]
target = 32                       # hari pertama yang lebih panas dari 32 °C?

ketemu = None                     # None = "belum ketemu"
cek = 0
for i in range(len(suhu)):
    cek = cek + 1
    if suhu[i] > target:
        ketemu = i
        break                     # berhenti di kecocokan pertama

print("Hari pertama di atas", target, "°C: indeks", ketemu)
print("Elemen yang diperiksa:", cek, "dari", len(suhu))

# Jalan pintas yang berguna
print(33 in suhu)                 # apakah 33 ada di dalam list?
print(suhu.index(33))             # posisi 33 yang pertama (error jika tidak ada)
```

```python
# Penyaringan: list baru berisi elemen yang cocok

suhu = [28, 31, 30, 33, 29]

panas = []                        # list baru, awalnya kosong
for t in suhu:
    if t >= 30:
        panas.append(t)
print("Nilai panas :", panas)
print("Jumlahnya   :", len(panas))

# Untuk menyimpan nomor hari, bukan nilainya, telusuri berdasarkan posisi
hari_panas = []
for i in range(len(suhu)):
    if suhu[i] >= 30:
        hari_panas.append(i + 1)
print("Hari panas  :", hari_panas)

# Opsional: penyaringan yang sama dengan list comprehension (perulangan ringkas)
panas_lagi = [t for t in suhu if t >= 30]
print("Hasil sama?", panas_lagi == panas)
```

```python
# Tetangga: perubahan harian dan rata-rata bergerak

suhu = [28, 31, 30, 33, 29, 27, 32]

# Perubahan harian: mulai dari indeks 1 karena kita melihat i - 1
for i in range(1, len(suhu)):
    perubahan = suhu[i] - suhu[i - 1]
    print(f"hari {i + 1}: {suhu[i]} °C, perubahan {perubahan:+d}")

# Rata-rata bergerak 3 hari: jendela = 3 nilai yang berakhir di indeks i
rata_3 = []
for i in range(2, len(suhu)):
    jendela = suhu[i - 2 : i + 1]
    rata_3.append(round(sum(jendela) / 3, 1))

print("Rata-rata 3 hari:", rata_3)
print(len(suhu), "nilai ->", len(rata_3), "rata-rata")

# AKTIVITAS 6E

suhu = [28, 31, 30, 33, 29, 27, 32]

# Rata-rata bergerak 5 hari
rata_5 = []
for i in range(4, len(suhu)):
    jendela = suhu[i - 4 : i + 1]
    rata_5.append(round(sum(jendela) / 5, 1))

print("Rata-rata 5 hari:", rata_5)
print(len(suhu), "nilai ->", len(rata_5), "rata-rata")
```

```python
# Rekap produksi energi bulanan

import csv

# --- Membaca data (disediakan) -> array paralel ---
tanggal, energi, jam_operasi = [], [], []
with open("daily_production_synthetic.csv") as f:
    for baris in csv.DictReader(f):
        tanggal.append(baris["date"])
        energi.append(int(baris["energy_mwh"]))
        jam_operasi.append(int(baris["run_hours"]))

DAYA_TERPASANG_MW = 330
NAMA_HARI = ["Min", "Sen", "Sel", "Rab", "Kam", "Jum", "Sab"]  # 1 November 2026 = Minggu

# 1. Total, rata-rata, capacity factor bulanan
total = sum(energi)
cf = total / (DAYA_TERPASANG_MW * 24 * len(energi)) * 100
print(f"Energi November : {total:,} MWh  |  rata-rata {total / len(energi):,.0f} MWh/hari")
print(f"Capacity factor : {cf:.1f} %")

# 2. Hari kerja vs akhir pekan (nama hari berulang setiap 7 indeks)
kerja, libur = [], []
for i in range(len(energi)):
    if NAMA_HARI[i % 7] in ("Sab", "Min"):
        libur.append(energi[i])
    else:
        kerja.append(energi[i])
print(f"Rata-rata hari kerja {sum(kerja) / len(kerja):,.0f} vs akhir pekan {sum(libur) / len(libur):,.0f} MWh")

# 3. Hari tidak beroperasi penuh 24 jam
for i in range(len(energi)):
    if jam_operasi[i] < 24:
        print(f"{tanggal[i]} ({NAMA_HARI[i % 7]}): hanya {jam_operasi[i]} jam beroperasi, {energi[i]:,} MWh")

# 4. Grafik batang teks: satu blok = 100 MWh di atas 3.000 MWh (agar perbedaan terlihat)
print()
for i in range(len(energi)):
    print(f"{i + 1:>2} {NAMA_HARI[i % 7]} {'█' * ((energi[i] - 3000) // 100)} {energi[i]:,}")
```

```python
# AKTIVITAS 6I

suhu = [29, 31, 30, 33, 28, 32, 34]

# (1) slicing
print("3 hari pertama:", suhu[:3])
print("2 hari terakhir:", suhu[-2:])

# (2) append lalu pop
suhu.append(27)
print("Setelah append:", suhu)
dikeluarkan = suhu.pop()
print("Dikeluarkan dengan pop:", dikeluarkan)
print("Setelah pop:", suhu)

# (3) sorted tanpa mengubah suhu
terurut = sorted(suhu)
print("Versi urut:", terurut)
print("suhu asli tetap:", suhu)

# (4) rata-rata dan suhu tertinggi + hari ke berapa
total = 0
for s in suhu:
    total += s
rata = total / len(suhu)
print("Rata-rata:", round(rata, 2))

idx_max = 0
for i in range(len(suhu)):
    if suhu[i] > suhu[idx_max]:
        idx_max = i
print("Suhu tertinggi:", suhu[idx_max], "pada hari ke-", idx_max + 1)

# (5) list suhu di atas rata-rata
di_atas = []
for s in suhu:
    if s > rata:
        di_atas.append(s)
print("Di atas rata-rata:", di_atas)
```

```python
# List 2D: tabel nilai

nilai = [
    [80, 70, 90],     # baris 0: Ana
    [65, 75, 60],     # baris 1: Budi
    [90, 85, 95],     # baris 2: Cici
]

print(nilai[1])        # satu baris utuh
print(nilai[1][2])     # baris 1, kolom 2: ujian 3 milik Budi
print("baris:", len(nilai), " kolom:", len(nilai[0]))

# Total untuk setiap baris (setiap mahasiswa)
for b in range(len(nilai)):
    print("Mahasiswa", b + 1, "total:", sum(nilai[b]))

# Satu kolom: indeks yang sama di setiap baris (ujian 1 semua orang)
ujian1 = []
for baris in nilai:
    ujian1.append(baris[0])
print("Skor ujian 1:", ujian1)

angka = [1, 2, 2, 3]

# Salah: menghapus dari list saat menelusurinya
for x in angka:
    if x == 2:
        angka.remove(x)
print("salah:", angka)          # satu angka 2 masih tersisa!

# Benar: buat list baru hanya dengan yang ingin dipertahankan
angka = [1, 2, 2, 3]
sisa = []
for x in angka:
    if x != 2:
        sisa.append(x)
print("benar:", sisa)
```

```python
# LIMA LATIHAN KECIL

skor = [80, 65, 90, 75, 70, 85]

# (1) skor pertama dan terakhir
print("Pertama:", skor[0])
print("Terakhir:", skor[-1])

# (2) rata-rata
total = 0
for s in skor:
    total += s
rata = total / len(skor)
print("Rata-rata:", rata)

# (3) jumlah skor di atas rata-rata
jumlah = 0
for s in skor:
    if s > rata:
        jumlah += 1
print("Di atas rata-rata:", jumlah)

# (4) posisi skor terendah (mahasiswa pertama = 1)
idx_min = 0
for i in range(len(skor)):
    if skor[i] < skor[idx_min]:
        idx_min = i
print("Mahasiswa terendah nomor:", idx_min + 1)

# (5) list baru, setiap skor + 5
baru = []
for s in skor:
    baru.append(s + 5)
print("List baru:", baru)
```

```python
# Laporan produksi bulanan (portofolio)

# GANTI dengan 30 angka asli dari blok data di materi
energi = [42, 45, 40, 48, 50, 47, 44, 41, 39, 43,
          46, 49, 52, 51, 48, 45, 42, 38, 40, 44,
          47, 50, 53, 49, 46, 43, 41, 45, 48, 44]

n = len(energi)

# 1) Total dan rata-rata
total = 0
for e in energi:
    total += e
rata = total / n
print("Total energi:", total)
print("Rata-rata   :", round(rata, 2))

# 2) Tiga tertinggi dan tiga terendah
pasangan = []
for i in range(n):
    pasangan.append([energi[i], i + 1])
urut = sorted(pasangan)

print("\n3 hari terendah:")
for nilai, hari in urut[:3]:
    print("  Hari", hari, ":", nilai)
print("3 hari tertinggi:")
for nilai, hari in urut[::-1][:3]:
    print("  Hari", hari, ":", nilai)

# 3) Hari di bawah target
target = float(input("\nMasukkan target: "))
di_bawah = 0
for e in energi:
    if e < target:
        di_bawah += 1
print("Hari di bawah target:", di_bawah)

# 4) Rata-rata bergerak 7 hari < 97% rata-rata bulanan
batas = 0.97 * rata
print("\nRata-rata bergerak 7 hari di bawah", round(batas, 2), ":")
ada = False
for i in range(6, n):
    ma = sum(energi[i - 6 : i + 1]) / 7
    if ma < batas:
        print("  Hari", i + 1, ": MA =", round(ma, 2))
        ada = True
if not ada:
    print("  (tidak ada)")

# 5) Total per minggu + sisa hari
print("\nTotal energi per minggu:")
for w in range(4):
    mulai = w * 7
    print("  Minggu", w + 1, ":", sum(energi[mulai : mulai + 7]))
print("  Sisa hari (29-30):", sum(energi[28:]))

```
